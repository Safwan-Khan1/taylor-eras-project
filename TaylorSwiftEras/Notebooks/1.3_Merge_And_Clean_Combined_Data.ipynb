```python
import os
import pandas as pd

all_song_path = os.path.join(project_folder_path ,'Data/Processed/taylor_song_df_processed.csv' )
all_lyrics_path = os.path.join(project_folder_path, 'Data/Processed/taylor_lyric_df_processed.csv')

taylor_song_df = pd.read_csv(all_song_path)
taylor_lyrics_df = pd.read_csv(all_lyrics_path)
```

```python
import re

def normalize_title(x):
    x = str(x).lower()

    # remove anything inside brackets
    x = re.sub(r"\(.*?\)", "", x)

    # remove punctuation
    x = re.sub(r"[^a-z0-9\s]", "", x)

    # normalize spaces
    x = re.sub(r"\s+", " ", x)

    return x.strip()
```

```python
taylor_song_df["track_name_clean"] = taylor_song_df["track_name"].apply(normalize_title)

taylor_lyrics_df["song_title_clean"] = taylor_lyrics_df["song_title"].apply(normalize_title)
```

```python

# Merge only the lyric column from df_grouped
taylor_merged_df = taylor_song_df.merge(
    taylor_lyrics_df[["song_title_clean", "lyric"]],
    left_on="track_name_clean",
    right_on="song_title_clean",
    how="left"
)

# Remove duplicate title column
taylor_merged_df = taylor_merged_df.drop(columns=["song_title_clean" , "track_name_clean" , "lyrics"])

# Preview
taylor_merged_df
```

```python
sheet = sheets.InteractiveSheet(df=taylor_merged_df)
```

```python
taylor_merged_df.info()
```

```python
taylor_merged_df = taylor_merged_df.dropna(subset=['lyric'])

string_cols = taylor_merged_df.select_dtypes(include ='object').columns

for col in string_cols:
    taylor_merged_df[col] = taylor_merged_df[col].fillna(taylor_merged_df[col].mode()[0])

taylor_merged_df = taylor_merged_df.drop_duplicates(subset="track_name")

taylor_merged_df.info()
```
