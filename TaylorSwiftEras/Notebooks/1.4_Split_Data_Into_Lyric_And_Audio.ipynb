####1.4 Split the merged data into two part , lyrics and audio to ease the task

```python
from google.colab import drive

drive.mount('/content/drive')
project_folder_path = "/content/drive/MyDrive/TaylorSwiftEras/"
```

```python
import os
import pandas as pd

taylor_merged_path = os.path.join(project_folder_path ,'Data/Final/taylor_merged_df.csv' )
taylor_merged_df = pd.read_csv(taylor_merged_path)
taylor_merged_df
```

```python
taylor_lyric_df = taylor_merged_df[['album_name' , 'track_number' , 'track_name' , 'lyric']].copy()
taylor_lyric_df
```

```python
taylor_lyric_df.to_csv(os.path.join(project_folder_path ,'Data/Final/taylor_lyric_df.csv') , index=False)
print('Successfully added file : taylor_lyric_df.csv')
```

```python
taylor_audio_df = taylor_merged_df[['album_name' , 'track_number' , 'track_name' , 'danceability' , 'energy',
                                    'key' , 'loudness' , 'mode' , 'speechiness' , 'acousticness' , 'instrumentalness',
                                    'liveness' , 'valence' , 'tempo' , 'time_signature' , 'duration_ms' ,
                                    'explicit' , 'key_name' , 'mode_name' , 'key_mode']].copy()
taylor_audio_df
```

```python
taylor_audio_df.to_csv(os.path.join(project_folder_path ,'Data/Final/taylor_audio_df.csv') , index=False)
print('Successfully added file : taylor_audio_df.csv')
```
