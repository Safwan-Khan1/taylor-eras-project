###**1. COLLECTING TAYLOR SWIFT SONGS , ALBUM AND ERAS DATA**

####1.1 Downloading existing datasets from kaggle and saving it to local directory

```python
from google.colab import drive

drive.mount('/content/drive')
project_folder_path = "/content/drive/MyDrive/TaylorSwiftEras/"
```

```python
import kagglehub
import shutil
import os

# Download dataset
path : str = kagglehub.dataset_download(
    "joebeachcapital/taylor-swift-all-songs-and-albums"
)

print("Downloaded to:", path)

# See files inside dataset
print(os.listdir(path))
```

```python
source_file : str = os.path.join(path, "taylor_all_songs.csv")

destination_folder : str = os.path.join(project_folder_path , "Data/Raw")


destination_file : str = os.path.join(
    destination_folder,
    "taylor_all_songs.csv"
)

shutil.copy(source_file, destination_folder)

print("CSV copied to:", destination_folder)
```

```python
import pandas as pd

taylor_song_df : pd.DataFrame = pd.read_csv(destination_file)
taylor_song_df_processed : pd.DataFrame = taylor_song_df.copy()
taylor_song_df
```

```python
#Dropping null index and columns
taylor_song_df_processed : pd.DataFrame = taylor_song_df_processed.dropna(subset = ['album_name'])
taylor_song_df_processed : pd.DataFrame = taylor_song_df_processed.drop(columns = ['featuring' , 'bonus_track' , 'promotional_release' , 'single_release'])
```

```python
taylor_song_df_processed['artist'] = taylor_song_df_processed['artist'].fillna('Taylor Swift')
taylor_song_df_processed.info()
```

```python
taylor_song_df_processed['lyrics'] = taylor_song_df_processed['lyrics'].astype('string')
taylor_song_df_processed
```

```python
#Filling null value in all numeric type columns with mean
numeric_cols  = taylor_song_df_processed.select_dtypes(include = ['number']).columns

for cols in numeric_cols :
  taylor_song_df_processed[cols] = taylor_song_df_processed[cols].fillna(taylor_song_df_processed[cols].mean())

#Filling null value in string and object column with mode
string_cols = taylor_song_df_processed.select_dtypes(include = ['object' , 'string']).columns

for col in string_cols:
    taylor_song_df_processed[col] = taylor_song_df_processed[col].fillna(taylor_song_df_processed[col].mode())

taylor_song_df_processed
```

```python
#Done fill all the null value in the dataframe
taylor_song_df_processed.to_csv(os.path.join(project_folder_path ,'Data/Processed/taylor_song_df_processed.csv' ), index = False)
```
