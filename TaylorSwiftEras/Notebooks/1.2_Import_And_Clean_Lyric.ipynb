```python
import kagglehub
import shutil
import os

# Download latest version
path = kagglehub.dataset_download("pradhanmanva/taylorswiftlyrics")

print("Path to dataset files:", path)
print(os.listdir(path))
```

```python
source_file : str = os.path.join(path, "final_taylor_swift_lyrics.tsv")

destination_folder : str = os.path.join(project_folder_path , "Data/Raw")


destination_file : str = os.path.join(
    destination_folder,
    "final_taylor_swift_lyrics.tsv"
)

shutil.copy(source_file, destination_folder)

print("CSV copied to:", destination_folder)
```

```python
import pandas as pd
import csv

taylor_lyric_df = pd.read_csv(
    destination_file,
    sep="\t",
    engine="python",
    quoting=csv.QUOTE_NONE,
    on_bad_lines="skip"
)
taylor_lyric_df
```

```python
bad_keywords = ["voice memo", "poem", "commentary", "interview"]

mask = ~taylor_lyric_df["song_title"].str.lower().str.contains(
    "|".join(bad_keywords),
    na=False
)

taylor_lyric_df = taylor_lyric_df[mask]
```

```python
taylor_lyric_df = taylor_lyric_df.groupby(["album","song_title"])["lyric"].apply(lambda x: " ".join(x)).reset_index()
taylor_lyric_df
```

```python
taylor_lyric_df.to_csv(os.path.join(project_folder_path , "Data/Processed/taylor_lyric_df_processed.csv"))
```
